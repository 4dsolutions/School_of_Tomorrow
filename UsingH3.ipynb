# H3 in Action

Of interest was to experiment with H3, the open mapping system based on a hexapent, a hexagons + 12 pentagons mosaic covering Planet Earth at varying levels of resolution. Around here, we'd call that a Global Matrix in the years leading up to this specific artifact's development.

The project: get a lat / long reading for School of Tomorrow HQS and use that to pull up and display the containing H3 geotile.

The workflow:

Start with Google Earth and dial in the location using the already-on-file street address:

Start out in Google colab and prompt AI for a skeletal script, from which scratch pad Notebook the following code was obtained and executed.

<a data-flickr-embed="true" href="https://www.flickr.com/photos/kirbyurner/55604411895/in/dateposted-public/" title="Using H3 from Google Colab"><img src="https://live.staticflickr.com/65535/55604411895_b45ac10d4d_z.jpg" width="640" height="381" alt="Using H3 from Google Colab"/></a><script async src="//embedr.flickr.com/assets/client-code.js" charset="utf-8"></script>

```python
    import h3
    
    lat = 45.5089625
    lng = -122.6247132
    resolution = 9
    
    h3_address = h3.latlng_to_cell(lat, lng, resolution)
    print(f"H3 Hexagon Address (Resolution {resolution}): {h3_address}")
```

The printed output: 

H3 Hexagon Address (Resolution 9): 8928f0030afffff

A subsequent attempt to lookup and display the tile from within colab hit a roadblock: OpenStreet Map blocked access.

I prompted Google Gemini (at the Google search prompt) for more clues and it recommended using H3's website.

Plugging in this tile address, I was able to pull up the requisite geotile using the Uber H3 website for that purpose:

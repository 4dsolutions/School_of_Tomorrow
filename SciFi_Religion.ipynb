# Preliminary Report on the Gospel of the Joined Faces
### Codex of the Southern Cistern: archaeological description, provisional translation, and interpretive notes

*An imagined report by a twenty-first-century archaeologist and antiquarian. The discovery, manuscript, translations, and field observations are fictional; the comparisons with historical writings are identified as such.*

## I. The discovery

I have provisionally named the manuscript *The Gospel of the Joined Faces*. Its surviving title, translated more literally, is *The Good Tidings of Those Who Become a Chamber*. The difference is not merely stylistic. What its author means by salvation appears to depend upon the distinction between a surface and an enclosure.

We recovered the codex from a disused cistern beneath a small settlement on the eastern margin of the Egyptian desert. It lay inside a ceramic vessel containing neither coins nor the usual documentary miscellany, but a copper mirror, four bone rods of equal length, and a folded sheet bearing a geometrical construction. Two further rods were discovered beneath the vessel. Whether these six pieces constituted a teaching apparatus cannot yet be established.

The manuscript comprises thirty-one surviving leaves. Its principal text is Coptic, with Greek technical terms retained in several passages. A second hand has added diagrams and brief objections in the margins. The binding incorporates an administrative fragment whose date, if genuine and correctly read, would place the binding no earlier than the late fourth century. That is not a date for the composition. Radiocarbon analysis, ink characterization, and a full palaeographical assessment remain pending.

I emphasize these limitations because the document invites precisely the kind of premature certainty against which an excavator must defend himself.

Its most disquieting assertion occurs on the seventh surviving leaf:

> The disciples said to him: “What shall we do with vanity, and where shall we cast hatred?”
>
> The Living One said: “Who taught you to cast away the metals before you had learned the furnace?”

## II. A gospel against spiritual waste

The text takes the form of conversations between an unnamed revealer, usually called “the Living One,” and several interlocutors. Among the latter are Mariam, a dyer, and a man designated only as “the measurer.” Whether these are imagined historical persons, ritual offices, or personifications is uncertain.

I call the work gnostic provisionally. Its saving knowledge concerns an ignorance deeper than failure to assent to doctrine: human beings do not know how they are assembled. Its rulers maintain their dominion by keeping the soul’s powers separate and persuading each that it is the whole.

Yet the manuscript complicates its own apparent dualism. The material world is not simply a prison to be abandoned. It is also a workshop whose equipment has been misunderstood.

> The ruler did not make your hunger.
>
> He taught your hunger to call itself king.

Likewise, the passions are not foreign substances implanted in an originally innocent soul. They are powers exercising an illegitimate sovereignty.

Vanity says: *Only I must be seen.*

Hatred says: *What wounds me must cease to exist.*

Love does not appear when these powers are merely added together. That combination, the author observes with considerable psychological accuracy, ordinarily produces persecution.

Instead, each must surrender its claim to govern the whole while retaining something of its original strength.

> Do not break the mirror. Turn it.
>
> Do not blunt the sword. Ask what it guards.

The gospel’s central proposition is therefore more exacting than “evil produces good.” It proposes that capacities imprisoned in vicious relationships can enter a different order of relationship—and that the resulting whole may possess a quality absent from its components in isolation.

## III. The furnace and its contents

The alchemical vocabulary is pervasive, although the surviving text contains no reproducible recipe for metallic transmutation. Its vessels are at once utensils, bodies, and communities.

One passage reads:

> Vanity is the silver that desires every face to be its own.
>
> Hatred is the fire that desires every difference to become ash.
>
> Place neither upon the throne.
>
> Let the silver receive the face of the stranger.
>
> Let the fire keep the stranger warm.
>
> Then the vessel shall contain what neither metal nor fire possessed.

Here vanity’s appetite for recognition becomes an ability to recognize. Hatred’s concentrated refusal becomes resistance to that which injures another. Neither transformation excuses the original vice. The author repeatedly distinguishes the preservation of a capacity from the preservation of its habitual object.

The closest historical comparison is not with a modern manual of positive thinking but with the visionary alchemy associated with Zosimos of Panopolis. His *Visions* bring material operations into relation with spiritual transformation, within a setting that combines Gnostic, Hermetic, and Platonic elements. That establishes a plausible comparative environment, not evidence that our author read him. [ora.ox.ac](https://ora.ox.ac.uk/objects/uuid:45fed8a7-0777-4822-b4a2-0fcc788d0b64)

Our manuscript’s distinctive contribution is its insistence that the furnace cannot be operated alone.

> Whoever heats himself in a sealed vessel without a witness becomes enamored of his own smoke.

The witness does not merely encourage the practitioner. She contradicts him. In one dialogue, Mariam refuses the measurer’s claim that his cruelty has become righteous indignation:

> “You have changed the inscription on the vessel,” she said, “but its contents still burn the servant.”

This is an important safeguard against a permissive reading. The gospel is not granting the initiate a license to rename his appetites.

## IV. The geometry of conversion

The diagrams initially appeared to be ornamental: triangles, intersecting circles, and arrangements of points. The most consequential drawing, however, depicts four triangular faces around an enclosed space.

Beside it stands the instruction:

> Four faces, and none is the chamber.
>
> Six meetings, and none is the guest.
>
> Seek what comes to dwell when the faces cease to lie apart.

The four faces are labelled, in my provisional translation:

- The desire to appear.
- The power to refuse.
- The endurance of another’s testimony.
- The remembrance of one’s own wound.

The first two correspond to vanity and hatred. The other two prevent their conjunction from becoming tyranny.

The diagram is thus not adequately represented by the formula “vanity plus hatred equals love.” Its claim is that self-regard and aversion, brought into a disciplined relationship with another person’s reality and one’s own vulnerability, can participate in a new psychological structure.

Love is the chamber, not one of the faces.

An adjacent gloss makes the distinction especially clear:

> If you write LOVE upon one face, you will worship a wall.

I find this the most intellectually formidable sentence in the codex. It refuses to identify love with a single privileged feeling. Tenderness may be one of its manifestations; so may protective anger. But no manifestation is entitled to declare itself the entire structure.

The accompanying ritual appears to have required participants to hold the rods together. No individual could complete the figure with one hand.

I cannot yet determine whether this is practical pedagogy or a liturgical instruction. The distinction may have been unintelligible to its users.

## V. A disagreement with the sculptor

The manuscript’s relationship to Platonism is intimate but disputatious.

Plotinus’s familiar image in *Enneads* I.6.9 instructs the aspirant to work upon the self as a sculptor works upon a statue: removing excess, straightening crookedness, and bringing obscured beauty to light. [loebclassics](https://www.loebclassics.com/abstract/plotinus-enneas/1966/pb_LCL440.259.xml?readMode=recto)

Our author seems to know a related spiritual vocabulary, although direct dependence cannot be demonstrated. One dialogue has the measurer say:

> “I shall cut away what is shameful until only the beautiful remains.”

Mariam answers:

> “And when you have cut away the hand that trembles, who will touch the frightened?”

The disagreement is not necessarily with purification itself. It concerns what purification means.

For the sculptor, the governing image is subtraction. For the author of the *Joined Faces*, it is rearrangement under constraint.

The latter asks whether the shameful trait conceals a capacity whose destruction would impoverish the person. The wish to be admired might conceal a wish to participate in beauty. Anger might conceal the recognition that something ought not be endured. Possessiveness might contain an undeveloped ability to remain faithful.

These are possibilities, not automatic vindications. The text is emphatic that a passion’s intensity does not prove its truth.

Its geometric ascent is correspondingly unusual. The soul does not rise by making itself ever thinner, until nothing earthly remains. It rises by acquiring relationships it can sustain without either consuming or being consumed.

> The point said, “I am without division.”
>
> The chamber answered, “You are also without a guest.”

## VI. Reading the codex after Fuller

At this point I must disclose a prejudice belonging to the excavator rather than the excavation.

I came to this manuscript having read **Buckminster Fuller**.

Fuller’s *Synergetics* treats the tetrahedron as fundamental to a system of spatial relations; his account of synergy concerns whole-system behavior not predictable from the behavior of components considered separately. His geometric framework emphasizes triangular and tetrahedral coordination rather than treating the cube as the unquestioned starting point. [en.wikipedia](https://en.wikipedia.org/wiki/Synergetics_(Fuller))

It would be irresponsible to announce that a late-antique scribe anticipated Fuller’s system. Nothing in our codex warrants attributing his mathematical program, terminology, or empirical ambitions to its author.

Nevertheless, Fuller has given me a useful question to ask of the drawings: not *What does each piece symbolize?* but *What becomes possible only when the pieces constitute a system?*

The distinction matters.

A collection of individually admirable traits does not necessarily make a loving person. Courage, intelligence, and discipline may cooperate in an atrocity. Conversely, a person’s shame, wounded pride, or anger may enter a relationship that changes what those powers do.

The manuscript’s geometry makes this relational claim visible. A face has properties; an enclosure has additional properties that no isolated face possesses. The analogy is suggestive, not a mathematical proof of moral psychology.

The text itself appears to understand that limitation:

> The figure does not make you merciful.
>
> It shows you where mercy cannot stand if you remove the other.

I have also found Fuller's vocabulary of tension and compression useful as a modern interpretive aid. His work relates these complementary structural conditions within its broader account of systems.  Here, too, I must resist importing the analogy as though it were an ancient technical doctrine. [bfi](https://www.bfi.org/about-fuller/big-ideas/synergetics/)

What the gospel actually supplies is more modest, and perhaps more humane: an argument that opposition need not disappear for a whole to become coherent.

> Do not command the bow to have no tension.
>
> Command the archer to answer for the arrow.

## VII. The ethical difficulty

The most dangerous possible interpretation would be that hatred deserves cultivation because love requires it.

The manuscript does not say this.

Its author distinguishes between finding a use for damaged material and deliberately producing damage. A particularly severe passage answers a disciple who proposes to sin so that he may possess more substance for transformation:

> Does the physician break the child’s leg because he knows the setting of bones?

Nor does the work suggest that victims must remain available for the spiritual education of those who harm them.

One damaged but legible injunction reads:

> The wounded one owes no furnace to the hand that struck.

This considerably changes my understanding of its communal practices. The “witness” cannot simply be an injured person compelled to validate the offender’s conversion. Her refusal must remain possible, or the supposed chamber is merely another prison.

The criterion of successful transmutation is therefore neither ecstasy nor the initiate’s confidence in his own enlightenment. It is a change in what becomes possible for others.

> If your gold cannot ransom anyone, inquire whether it is gold.

By “ransom,” the surrounding discussion seems to mean concrete release: from debt, coercion, humiliation, or fear. The gospel’s esotericism does not exempt it from consequences. Rather, consequences expose counterfeit knowledge.

## VIII. The leaf behind the mirror

The copper mirror was originally catalogued separately from the manuscript. During conservation, a folded scrap became visible behind its backing.

It contains no revelation, genealogy of heavenly powers, or diagram of ascent. It records names followed by quantities of grain.

One name has been crossed out and entered again beneath a heading I translate, provisionally, as “What is owed to her.”

The same hand appears in a marginal note beside the teaching on vanity:

> I wished to be praised for returning it.
>
> I returned it.
>
> The wish remained.
>
> She ate.

I have spent more time considering these four lines than any other passage in the find.

They prevent the gospel from dissolving into a beautiful theory of eventual perfection. Here is an individual who does not claim to have eliminated vanity before acting justly. Nor does he claim that justice has magically purified his motive. He records an imperfect action that nevertheless altered another person’s circumstances.

Love, on this account, need not wait for immaculate ingredients.

But neither is it merely a flattering name for their mixture.

## IX. Provisional conclusion

The *Gospel of the Joined Faces* presents salvation as a change in organization: the soul ceases to be a succession of rival claimants and becomes capable of sheltering someone other than itself.

Its alchemy supplies the language of transformation. Its geometry supplies a discipline of relationship. Its gnostic revelation is the discovery that what has ruled us need not be destroyed in every respect, but must cease to rule.

My familiarity with Fuller undoubtedly makes me more attentive to the possibility that the whole possesses capacities absent from the isolated parts. The archaeological task is to keep that illumination from becoming an anachronism.

I therefore conclude with the manuscript rather than with my interpretation:

> They asked him: “When shall we know that the work is finished?”
>
> He said: “When you no longer ask the stranger to admire the house before you let him enter.”

Below this, in the smaller hand responsible for the objections, someone has written:

> And when he may leave.

## Response

Your paragraph is broadly on the right track and contains a genuinely interesting insight (sense-making as prior to truth), but it also compresses several complex shifts in 20th‑century philosophy into a single, slightly misleading narrative. It’s perspicuous as a first sketch, but it would benefit from sharpening some distinctions and correcting a few historical/conceptual conflations.

## What’s accurate and insightful

- **Linguistic Turn as focus on language.** It’s right that “the Linguistic Turn” (a phrase popularized by Rorty, coined by Bergmann) designates a widespread shift toward treating philosophical problems as problems about language: how our concepts are embodied in linguistic usage and how clarifying that usage can dissolve or resolve philosophical puzzles. [rep.routledge](https://www.rep.routledge.com/articles/thematic/linguistic-turn/v-1)

- **Wittgenstein and sense vs. nonsense.** Your characterization of Wittgenstein—especially as concerned with *sense* and *nonsense*—is very much in line with standard readings. In the *Tractatus*, the aim is explicitly “to distinguish between sense and nonsense,” by finding the limits of what can be expressed in language. Later, in *Philosophical Investigations*, he shifts from “meaning as representation” to “meaning as use”, and nonsense becomes a matter of language going on holiday, i.e. misfiring within our practices. [plato.stanford](https://plato.stanford.edu/entries/wittgenstein/)

- **Sense prior to truth.** The idea that sense precedes truth is philosophically respectable and textually grounded. Frege and (early) Wittgenstein both treat *sense* as the condition of something even being evaluable as true or false; one can grasp the sense of a proposition without yet knowing its truth value. So your line “For thinking to be true, it first has to make sense” nicely condenses a deep point about the logical order: no truth or falsity without prior intelligibility. [plato.stanford](https://plato.stanford.edu/entries/wittgenstein/)

- **Philosophy’s scope widening.** It’s plausible to say that attention to language and sense broadened philosophy’s remit: instead of just asking “is this claim true?”, philosophers increasingly asked “does this claim even say anything, and how does it function in our language-game?”. The linguistic turn, especially in its ordinary-language variant, put many previously “metaphysical” questions under a new light: are they genuine questions or verbal illusions? [rep.routledge](https://www.rep.routledge.com/articles/thematic/linguistic-turn/v-1)

## Where the picture is too simplified

### 1. “Before the Linguistic Turn: true vs. false”

You write as if the *primary* focus of philosophy before the linguistic turn was “true versus false” and securing a “rock solid crystalline substrate” of rationality. That matches *some* strands (e.g. foundationalist epistemology, rationalist metaphysics, certain forms of logicism), but:

- Classical metaphysics and ethics often care more about *what there is* or *what is good* than about truth-values per se.
- Ancient and early-modern philosophy already considers meaning, understanding, and the conditions of intelligibility (e.g. Aristotle on categories, Kant on the conditions of possible experience, Hegel on conceptual development).

So: the contrast “truth vs. falsehood” → “sense vs. nonsense” is suggestive, but historically too stark. Better: the linguistic turn **recentered** many debates around language and sense, rather than replacing a previous exclusive focus on truth.

### 2. Linguistic Turn ≠ simple shift from truth to sense

The standard account of the linguistic turn is methodological: philosophy becomes “an a priori conceptual discipline” that elucidates the interrelations of concepts as embodied in language, in order to dispel confusions. This involves: [rep.routledge](https://www.rep.routledge.com/articles/thematic/linguistic-turn/v-1)

- Analytic focus on logic and ideal languages (Frege, Russell, Carnap, early Wittgenstein).
- Ordinary language focus on everyday usage (later Wittgenstein, Austin, Ryle).

Truth remains central for huge parts of analytic philosophy after the linguistic turn (truth-conditional semantics, Tarski, Davidson, Kripke, contemporary theories of truth). What changes is **how** truth, meaning, and representation are investigated: through logical form, semantics, and use, rather than via direct metaphysical intuition.

So your “true vs. false” → “sense vs. nonsense” captures an important *Wittgensteinian* strand, but it shouldn’t be generalized as *the* effect of the linguistic turn on philosophy as a whole.

### 3. Wittgenstein’s role

You’re right that Wittgenstein “gets a lot of the credit,” but the genealogy is more nuanced:

- The linguistic turn is commonly traced to *Tractatus* and then developed by logical positivists, who use his ideas to reject metaphysics and focus on the logical syntax of language. [rep.routledge](https://www.rep.routledge.com/articles/thematic/linguistic-turn/v-1)
- The **later** Wittgenstein then turns away from ideal-language projects, emphasizing ordinary language, meaning-as-use, language-games, and conceptual therapy. [plato.stanford](https://plato.stanford.edu/entries/wittgenstein/)

Your paragraph fuses these two phases into one, and leans heavily on the later work (“philosophical investigations were more about how words mean than about proving theses”), which is accurate for *PI* but not for the methodological role of the *Tractatus* in early analytic philosophy.

## Clarifying the “sense prior to truth” idea

You have a compelling core thesis: “the dimension of sense making inserted itself as prior to, more primitive than, that of truth telling.” Philosophically, this can be sharpened:

- A **proposition** must have *sense* (Sinn) to be truth-apt—i.e. capable of being true or false. [plato.stanford](https://plato.stanford.edu/entries/wittgenstein/)
- Logical positivists and Wittgenstein use this to treat many traditional metaphysical and theological sentences as *nonsense*—not merely false, but lacking the conditions under which they could be true or false. [rep.routledge](https://www.rep.routledge.com/articles/thematic/linguistic-turn/v-1)
- Later Wittgenstein extends this by treating much philosophy as a kind of *grammatical confusion*: sentences that look meaningful but lack a clear role in our language-games. [plato.stanford](https://plato.stanford.edu/entries/wittgenstein/)

So the move isn’t: “philosophy stopped caring about truth.” It’s: **philosophy began asking, more systematically, whether its own sentences have determinate sense and what their role in our practices is, before worrying about their truth.**

## Suggested tightened version

If you want a more perspicuous formulation that keeps your insight but cleans up the historical/conceptual edges, something like this would be closer to the mark:

> Before what we now call the “linguistic turn” in 20th‑century philosophy, much work—especially in metaphysics and epistemology—aimed at securing indubitable foundations for rational belief and knowledge, often framed in terms of truth and justification. With the linguistic turn (early analytic philosophy, logical positivism, and later ordinary‑language philosophy), a central strand of philosophy reconceived its task: not as discovering extra‑linguistic foundations, but as clarifying the concepts embodied in our language.  
>  
> Within this strand, especially in Wittgenstein, the question “Is this statement true?” is preceded by “Does this statement have a clear sense at all, given the way our language works?” For a sentence to be true or false, it must first be intelligible; many traditional philosophical sentences turn out, on this view, to be neither true nor false but nonsensical—misuses of language rather than incorrect descriptions of reality. [rep.routledge](https://www.rep.routledge.com/articles/thematic/linguistic-turn/v-1)

From that vantage point, your paragraph is conceptually rich and broadly aligned with Wittgensteinian motivations, but it overgeneralizes from that to “philosophy” as such and compresses a multi-phase, multi-tradition phenomenon (early vs later Wittgenstein, positivists vs ordinary language, ongoing truth‑theoretic work) into a single axis of “truth” vs “sense.”

## Response

The passage is **quite perspicuous rhetorically**: its central intuition—that questions of meaning or intelligibility can constrain questions of truth—is easy to grasp. But as intellectual history and as an account of Wittgenstein, it is too neat and needs several qualifications.

## Main corrections

- Richard Rorty **popularized** “the linguistic turn” through his 1967 anthology, but he credited Gustav Bergmann with coining the expression. [plato.stanford](https://plato.stanford.edu/archives/win2021/entries/rorty/)
- Philosophy before the linguistic turn was not chiefly a contest of “true versus false.” It also concerned being, causation, virtue, political order, consciousness, God, knowledge, and many other matters. The “rock-solid substrate” better describes a particular *foundationalist* aspiration in early modern epistemology and, later, certain analytic projects—not philosophy as such.
- The linguistic turn did not simply replace truth with sense. It made language, meaning, reference, logical form, and use central *methods* for addressing philosophical problems. In much analytic philosophy, semantic clarification was meant to help determine what could intelligibly be asserted, and hence what could be true or false. [rep.routledge](https://www.rep.routledge.com/articles/thematic/linguistic-turn/v-1)
- “Sense-making is prior to truth-telling” is defensible if stated carefully: a declarative utterance ordinarily needs determinate content before it can be evaluated as true or false. But this is a logical/semantic dependency, not necessarily a historical discovery of twentieth-century philosophy—Aristotle and medieval philosophers, among many others, distinguished meaningful assertion from mere noise or category error.

## Wittgenstein
|
Your formulation fits **early Wittgenstein** unusually well. In the *Tractatus*, propositions with sense are “bipolar”: they portray possible states of affairs and can therefore be true or false; philosophical pseudo-propositions may fail to say anything truth-evaluable at all. Its project explicitly includes marking a boundary between sense and nonsense. [plato.stanford](https://plato.stanford.edu/entries/wittgenstein/)

But *Philosophical Investigations* makes a rather different point. Rather than offering a general account of “how words mean,” it directs us to the diverse practices—“language-games”—in which words have use. This often dissolves philosophical temptations generated by detaching words from their ordinary roles. So Wittgenstein is not principally substituting semantic theory for metaphysics; he is often resisting the urge for a single philosophical theory altogether.

## Scope

“The scope of philosophy was thereby widened” is the least persuasive line. One could instead say that the **methodological agenda shifted**: philosophy came to treat linguistic and conceptual clarification as indispensable, sometimes even as philosophy’s distinctive task. For logical positivists and early Wittgenstein, this could *narrow* the range of legitimate philosophical assertion by classifying much traditional metaphysics as nonsensical or non-cognitive.

## A tighter version

> The linguistic turn did not replace truth with sense, but made questions of meaning, reference, logical form, and linguistic use central to philosophical method. On this view, before asking whether a purported assertion is true or false, one must ask whether it expresses a determinate proposition at all. Early Wittgenstein dramatized this point: propositions with sense are capable of truth or falsity, whereas some traditional philosophical utterances fail to state facts and therefore are not candidates for either. Later Wittgenstein shifted attention from logical form to the varied uses of words in human practices. Rorty popularized the label “linguistic turn,” though he attributed the phrase itself to Gustav Bergmann. [plato.stanford](https://plato.stanford.edu/archives/win2021/entries/rorty/)

This preserves your attractive “sense before truth” insight while avoiding a misleading before/after story.

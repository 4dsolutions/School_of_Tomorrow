```python
from tetravolume import root5
    
𝛟  = PHI

PD_coords = {'a':(0 , 𝛟 , 1/𝛟 , root5), 
             'b':(𝛟 , 0 , root5 , 1/𝛟), 
             'c':(1/𝛟 , root5 , 0 , 𝛟),
             'd':(root5 , 1/𝛟, 𝛟 , 0), 
             
             'e':(𝛟 , 1/𝛟 , 0 , root5),
             'f':(0 , root5 , 𝛟 , 1/𝛟),
             'g':(root5 , 0 , 1/𝛟 , 𝛟),
             'h':(1/𝛟 , 𝛟 , root5 , 0), 
             
             'i':(1/𝛟 , 0 , 𝛟 , root5),
             'j':(root5 , 𝛟 , 0 , 1/𝛟), 
             'k':(0 , 1/𝛟 , root5 , 𝛟),
             'l':(𝛟 , root5 , 1/𝛟 , 0),
             
             'm':(2, 2, 2, 0),
             'n':(2, 2, 0, 2),
             'o':(2, 0, 2, 2),
             'p':(0, 2, 2, 2),
    
             'q':(0, 0, 0, 2),
             'r':(0, 0, 2, 0),
             's':(0, 2, 0, 0),
             't':(2, 0, 0, 0)}

# checks out
IC_coords =  {'a':(0 , 1 , 𝛟 , 𝛟**2),
              'b':(1 , 0 , 𝛟**2 , 𝛟),
              'c':(𝛟 , 𝛟**2 , 0 , 1),
              'd':(𝛟**2 , 𝛟, 1 , 0 ),
              'e':(1 , 𝛟 , 0 , 𝛟**2),
              'f':(0, 𝛟**2 , 1 , 𝛟 ),
              'g':(𝛟**2 , 0 , 𝛟 , 1),
              'h':(𝛟 , 1 , 𝛟**2 , 0),
              'i':(𝛟 , 0 , 1 , 𝛟**2),
              'j':(𝛟**2 , 1 , 0 , 𝛟),
              'k':(0 , 𝛟 , 𝛟**2 , 1),
              'l':(1 , 𝛟**2 , 𝛟 , 0)}
    
```

```python
for v in pd.vertexes:
    a = float(pd.vertexes[v].a.evalf(5))
    b = float(pd.vertexes[v].b.evalf(5))
    c = float(pd.vertexes[v].c.evalf(5))
    d = float(pd.vertexes[v].d.evalf(5))
    print(f"({a:.4f},{b:.4f},{c:.4f},{d:.4f})") 
```

```python
faces = (('e','f','g'),('e','g','h'),
                      ('e','h','f'),('f','h','g'))
```

```python
unique = set()
for f in faces:
    for pair in zip(f , f[1:] + (f[0],)):
        unique.add( tuple(sorted(pair)) )
```
